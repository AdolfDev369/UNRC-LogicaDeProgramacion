```python
#PRACTICA 9
"Escribir un programa que pida al usuario el peso en kg y su estartura y calcula su indice de masa corporal"

#opcion 1
peso=float(input("Introduce tu peso en Kg "))
estatura=float(input("Introduce tu estatura en m "))
IMC=round(peso/(estatura*estatura))

print(f"Tu indice de masa corporal es ", IMC)

#opcion 2 esto round REDONDEA LOS DECIMALES.
peso=input("Introduce tu peso en Kg ")
estatura=input("Introduce tu estatura en m ")
IMC=round(float(peso)/float(estatura)**2)

print(f"Tu IMC es ", IMC)

```
